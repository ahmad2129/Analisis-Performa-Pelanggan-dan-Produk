### Outlier

#### Label Encoder

```python
le = LabelEncoder()
```

```python
df_copy_2 = df_copy.copy()
```

```python
'''Membuat Label Encoder Pada Kolom Dengan Tipe Data String'''
df_copy_2["Order ID Code"] = le.fit_transform(df_copy_2["Order ID"])
df_copy_2["Order Date Code"] = le.fit_transform(df_copy_2["Order Date"])
df_copy_2["Ship Date Code"] = le.fit_transform(df_copy_2["Ship Date"])
df_copy_2["Ship Mode Code"] = le.fit_transform(df_copy_2["Ship Mode"])
df_copy_2["Customer ID Code"] = le.fit_transform(df_copy_2["Customer ID"])
df_copy_2["Customer Name Code"] = le.fit_transform(df_copy_2["Customer Name"])
df_copy_2["Segment Code"] = le.fit_transform(df_copy_2["Segment"])
df_copy_2["Country Code"] = le.fit_transform(df_copy_2["Country"])
df_copy_2["City Code"] = le.fit_transform(df_copy_2["City"])
df_copy_2["State Code"] = le.fit_transform(df_copy_2["State"])
df_copy_2["Region Code"] = le.fit_transform(df_copy_2["Region"])
df_copy_2["Product ID Code"] = le.fit_transform(df_copy_2["Product ID"])
df_copy_2["Category Code"] = le.fit_transform(df_copy_2["Category"])
df_copy_2["Sub-Category Code"] = le.fit_transform(df_copy_2["Sub-Category"])
df_copy_2["Product Name Code"] = le.fit_transform(df_copy_2["Product Name"])
```

#### Cek Outlier Dengan IQR

```python
df_copy_outlier = df_copy_2[[
    "Order ID Code",
    "Order Date Code",
    "Ship Date Code",
    "Ship Mode Code",
    "Customer ID Code",
    "Customer Name Code",
    "Segment Code",
    "Country Code",
    "City Code",
    "State Code",
    "Region Code",
    "Product ID Code",
    "Category Code",
    "Sub-Category Code",
    "Product Name Code",
    "Row ID",
    "Postal Code",
    "Quantity",
    "Sales",
    "Discount",
    "Profit"
]]
```

```python
kolom_outlier = df_copy_outlier.columns
```

```python
for i in kolom_outlier :
    q1 = df_copy_outlier[i].quantile(0.25)
    q3 = df_copy_outlier[i].quantile(0.75)
    iqr = (q3 - q1)
    upper = q3 + (iqr*1.5)
    lower = q1 - (iqr*1.5)
    outlier = (df_copy_outlier[i] < lower) | (df_copy_outlier[i] > upper)
    jumlah_outlier = outlier.sum()
    persentase_outlier = jumlah_outlier/len(df_copy_outlier)*100
    print(f"Jumlah Outlier Kolom {i} = {jumlah_outlier}")
    print(f"Persentase Outlier Kolom {i} = {persentase_outlier:.2f}%\n")
```
